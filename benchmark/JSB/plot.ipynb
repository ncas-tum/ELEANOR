```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# df_felif_0005 = pd.read_pickle("review/FeLIF/0.0005/model_0.pkl")
df_felif_005 = pd.read_pickle("review/FeLIF/0.005/model_0.pkl")
df_bruno_005 = pd.read_pickle("review/Bruno/0.005/model_0.pkl")

# df_felif_0005 = df_felif_0005.assign(Variability=0.0005, Model="BPTT")
df_felif_005 = df_felif_005.assign(Variability=0.005, Model="BPTT")
df_felif_005 = df_felif_005.drop(columns=["Learning rate"])

df_bruno_005 = df_bruno_005.assign(Variability=0.005)
df_bruno_005 = df_bruno_005.drop(columns=["Learning rate"])


df = pd.concat([df_bruno_005, df_felif_005], ignore_index=True)
df = df.rename(columns={"Variability": "Learning rate"})
df = df.rename(columns={"Model": "Method"})

with mpl.style.context("boilerplot.ieeetran"):
    fig, ax = plt.subplots(figsize=[3.45, 2.3], dpi=200)
    sns.set_palette("Dark2")
    sns.lineplot(
        df, x="Epoch", y="Loss", style="Learning rate", hue="Method", ax=ax, linewidth=2
    )
    # sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
    plt.yscale("log")
    plt.xscale("log")
    # plt.savefig('jsb_methods.pdf')
    plt.show()
```
